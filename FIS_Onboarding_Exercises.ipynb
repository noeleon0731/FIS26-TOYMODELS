```python
import numpy as np
x=np.zeros([21,4])

for i in range(21):
  x[i][0]= i
  x[i][1]= i ** 2
  x[i][2]= i ** 3
  x[i][3]= i ** 4

print("columns:")
print(x[:,0])
print(x[:,1])
print(x[:,2])
print(x[:,3])
```

**Problem 5:** Print the values of i for which sin(i) is between -0.1 and +0.1 for $0<i<150$ radians.  *Hint*:  sin(i) is not built into python, but it is in numpy.  

```python
import numpy as np
import math as math

x=np.zeros([151,1])

for i in range(151):
  x[i][0] = math.sin(i)

target=0
tol=0.1
hits=np.array([])

print("target:")
print(target,"±",tol)
for val in x:
  if np.abs(val-target)<tol:
    hits=np.append(hits,val)

print("list of nearby values:")
print(hits)
```

**Problem 6:** Download the londonincidents data set. Use matplotlib to show a scatter plot of the "data" array stored above, with the first column on the x-axis and the second column on the y-axis.  Set the aspect ratio of your plot to be 1 (as wide as it is high).  

```python
import numpy as np  #load numpy
data=np.load("londonIncidents.npz") #load the data
data=data['all'] #extract the data from the archive
print(data) #print the data

import matplotlib.pyplot as plt

x=data[:,0]
y=data[:,1]
fig=plt.scatter(x,y)

ax=plt.gca()
ax.set_box_aspect(1)

plt.show()
```

**Problem 7:**  Sometimes you will want to plot different sets of data on the same graph.  In the following, values1 lists the x-value, y-value, and error bar for one dataset, and values2 show the x-value, y-value and error bar for another data set.  Plot them together, using different symbols and colors for each dataset, showing the error bar.  

```python
import matplotlib.pyplot as plt

values1=np.array([[0,1.2,0.3],[1,2.5,.2],[2,3.0,.4],[3,3.8,.3]])
values2=np.array([[0.5,1.0,0.4],[1.5,2.5,.3],[2.5,3.1,.2],[3.5,4.4,.3]])


plt.plot(values1[:,1],c='k',label='values1')
plt.plot(values2[:,1],c='r',label='values2')
plt.legend()
plt.show()
```
